```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, regexp_replace, to_date, coalesce, trim, upper, when

spark = SparkSession.builder.appName("ProyectoMedallion").getOrCreate()

# ==========================================
# CAPA BRONZE
# ==========================================
path_bronze = "/content/transacciones_bronze.csv"
df_bronze = spark.read.option("header", "true").csv(path_bronze)
print("--> Capa Bronze completada.")
df_bronze.printSchema()

# ==========================================
# CAPA SILVER
# ==========================================

# 1. Limpiar el monto: quitar $, comas, "MXN" y espacios, luego convertir a número
df_silver = df_bronze.withColumn(
    "monto",
    regexp_replace(
        regexp_replace(col("monto_str"), r"[\$,]", ""),   # quita $ y comas
        r"\s*MXN\s*", ""                                    # quita "MXN"
    ).cast("double")
)

from pyspark.sql.functions import expr

from pyspark.sql.functions import expr

from pyspark.sql.functions import coalesce

df_silver = df_silver.withColumn(
    "fecha",
    coalesce(
        to_date(col("fecha_operacion"), "yyyy-MM-dd"),
        to_date(col("fecha_operacion"), "dd/MM/yyyy"),
        to_date(col("fecha_operacion"), "MM-dd-yyyy"),
        to_date(col("fecha_operacion"), "dd-MMM-yyyy"),
        to_date(col("fecha_operacion"), "yyyy/MM/dd"),
    )
)

# 3. Limpiar categoría: quitar caracteres raros, acentos manuales y símbolos
df_silver = df_silver.withColumn(
    "categoria",
    trim(regexp_replace(col("categoria_raw"), r"[^a-zA-Z0-9áéíóúÁÉÍÓÚñÑ ]", ""))
)

# 4. Normalizar tipo de transacción a solo INGRESO / EGRESO
df_silver = df_silver.withColumn(
    "tipo_transaccion",
    when(upper(col("tipo_movimiento")).isin("ABONO", "INGRESO", "CREDITO", "CR"), "INGRESO")
    .when(upper(col("tipo_movimiento")).isin("CARGO", "EGRESO", "DEBITO", "DB"), "EGRESO")
    .otherwise(None)
)

# 5. Quedarnos solo con las columnas limpias (esquema Silver) y quitar filas rotas
df_silver = df_silver.select(
    col("id_transaccion"),
    col("id_usuario"),
    col("monto"),
    col("fecha"),
    col("categoria"),
    col("tipo_transaccion")
).na.drop(subset=["monto", "fecha", "tipo_transaccion"])

print("--> Capa Silver procesada correctamente:")
df_silver.show(10)
df_silver.printSchema()

print(f"Filas originales (Bronze): {df_bronze.count()}")
print(f"Filas limpias (Silver): {df_silver.count()}")
```

```python
print(f"Filas originales (Bronze): {df_bronze.count()}")
print(f"Filas limpias (Silver): {df_silver.count()}")
print(f"Filas descartadas por datos inválidos: {df_bronze.count() - df_silver.count()}")
```

```python
df_silver.toPandas().to_csv("/content/transacciones_silver.csv", index=False)
```

```python
from pyspark.sql.functions import sum as _sum, count, round, desc

# ==========================================
#  CAPA GOLD - REPORTES DE NEGOCIO
# ==========================================

# REPORTE 1: Balance General (Ingresos vs Egresos)
df_gold_balance = df_silver \
    .groupBy("tipo_transaccion") \
    .agg(
        round(_sum("monto"), 2).alias("total_monto"),
        count("id_transaccion").alias("total_operaciones")
    )

print("--- 1. Balance General (Ingresos vs Egresos) ---")
df_gold_balance.show()


# REPORTE 2: Gastos por Categoría (Solo EGRESOS)
df_gold_categorias = df_silver \
    .filter(col("tipo_transaccion") == "EGRESO") \
    .groupBy("categoria") \
    .agg(round(_sum("monto"), 2).alias("total_gastado")) \
    .orderBy(desc("total_gastado"))

print("--- 2. Top Categorías donde más se gasta ---")
df_gold_categorias.show(10)


# REPORTE 3: Resumen por Usuario (Monto total por cliente)
df_gold_usuarios = df_silver \
    .groupBy("id_usuario", "tipo_transaccion") \
    .agg(round(_sum("monto"), 2).alias("monto_total")) \
    .orderBy(desc("monto_total"))

print("--- 3. Resumen de Movimientos por Usuario (Top 10) ---")
df_gold_usuarios.show(10)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# ==========================================
# 1. EXPORTAR TABLAS GOLD A CSV
# ==========================================
# Guardamos los reportes para que los puedas descargar de Colab
df_gold_balance.coalesce(1).write.mode("overwrite").option("header", "true").csv("/content/gold_balance")
df_gold_categorias.coalesce(1).write.mode("overwrite").option("header", "true").csv("/content/gold_categorias")

print("--> Archivos de la Capa Gold guardados con éxito en la carpeta /content/\n")


# ==========================================
# 2. GRAFICAR RESULTADOS (VISUALIZACIÓN)
# ==========================================
# Convertimos las tablas de Spark a Pandas solo para graficar rápido
pdf_balance = df_gold_balance.toPandas()
pdf_categorias = df_gold_categorias.limit(5).toPandas() # Tomamos solo el top 5

# Crear figura con 2 gráficas
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfica 1: Ingresos vs Egresos (Dona / Pie)
ax1.pie(pdf_balance['total_monto'], labels=pdf_balance['tipo_transaccion'], autopct='%1.1f%%', startangle=90, colors=['#ff9999','#66b3ff'])
ax1.set_title('Distribución de Ingresos vs Egresos')

# Gráfica 2: Top 5 Categorías de Gasto (Barras)
ax2.barh(pdf_categorias['categoria'], pdf_categorias['total_gastado'], color='#ff9999')
ax2.set_title('Top 5 Categorías con Mayor Gasto')
ax2.set_xlabel('Monto Total ($)')
ax2.invert_yaxis() # Para mostrar el mayor gasto arriba

plt.tight_layout()
plt.show()
```
