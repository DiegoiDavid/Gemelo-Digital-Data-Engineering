```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, regexp_replace, to_date, coalesce, trim, upper, when

spark = SparkSession.builder.appName("ProyectoMedallion").getOrCreate()

# ==========================================
# CAPA BRONZE
# ==========================================
path_bronze = "/content/transacciones_bronze.csv"
df_bronze = spark.read.option("header", "true").csv(path_bronze)
print("--> Capa Bronze completada.")
df_bronze.printSchema()

# ==========================================
# CAPA SILVER
# ==========================================

# 1. Limpiar el monto: quitar $, comas, "MXN" y espacios, luego convertir a número
df_silver = df_bronze.withColumn(
    "monto",
    regexp_replace(
        regexp_replace(col("monto_str"), r"[\$,]", ""),   # quita $ y comas
        r"\s*MXN\s*", ""                                    # quita "MXN"
    ).cast("double")
)

from pyspark.sql.functions import expr

from pyspark.sql.functions import expr

from pyspark.sql.functions import coalesce

df_silver = df_silver.withColumn(
    "fecha",
    coalesce(
        to_date(col("fecha_operacion"), "yyyy-MM-dd"),
        to_date(col("fecha_operacion"), "dd/MM/yyyy"),
        to_date(col("fecha_operacion"), "MM-dd-yyyy"),
        to_date(col("fecha_operacion"), "dd-MMM-yyyy"),
        to_date(col("fecha_operacion"), "yyyy/MM/dd"),
    )
)

# 3. Limpiar categoría: quitar caracteres raros, acentos manuales y símbolos
df_silver = df_silver.withColumn(
    "categoria",
    trim(regexp_replace(col("categoria_raw"), r"[^a-zA-Z0-9áéíóúÁÉÍÓÚñÑ ]", ""))
)

# 4. Normalizar tipo de transacción a solo INGRESO / EGRESO
df_silver = df_silver.withColumn(
    "tipo_transaccion",
    when(upper(col("tipo_movimiento")).isin("ABONO", "INGRESO", "CREDITO", "CR"), "INGRESO")
    .when(upper(col("tipo_movimiento")).isin("CARGO", "EGRESO", "DEBITO", "DB"), "EGRESO")
    .otherwise(None)
)

# 5. Quedarnos solo con las columnas limpias (esquema Silver) y quitar filas rotas
df_silver = df_silver.select(
    col("id_transaccion"),
    col("id_usuario"),
    col("monto"),
    col("fecha"),
    col("categoria"),
    col("tipo_transaccion")
).na.drop(subset=["monto", "fecha", "tipo_transaccion"])

print("--> Capa Silver procesada correctamente:")
df_silver.show(10)
df_silver.printSchema()

print(f"Filas originales (Bronze): {df_bronze.count()}")
print(f"Filas limpias (Silver): {df_silver.count()}")
```

```python
print(f"Filas originales (Bronze): {df_bronze.count()}")
print(f"Filas limpias (Silver): {df_silver.count()}")
print(f"Filas descartadas por datos inválidos: {df_bronze.count() - df_silver.count()}")
```
